```python
import numpy as np
```

```python
def cleaning_policy(robot_pos, dirt_map):
  '''Greedy distance algorithm to find next closest dirt patch.'''
    r, c = robot_pos

    if dirt_map[r, c] == 1:
        return 4

    dirt_positions = np.argwhere(dirt_map == 1)

    if dirt_positions.size == 0:
        return np.random.randint(0, 4)

    distances = np.abs(dirt_positions[:, 0] - r) + np.abs(dirt_positions[:, 1] - c)

    nearest_index = np.argmin(distances)
    target_r, target_c = dirt_positions[nearest_index]

    if target_r < r:
        return 0
    elif target_r > r:
        return 2
    elif target_c > c:
        return 1
    elif target_c < c:
        return 3

    return np.random.randint(0, 4)
```
